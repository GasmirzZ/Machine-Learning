## 2. Nedostajuće vrednosti
### Pronalaženje
python
t.isnull().sum()

isnull() pravi tabelu istog oblika, punu True/False, a sum() ih broji po kolonama. Isti trik sa jedinicama i nulama koji već znaš.

Za procenat koristiš mean umesto sum:

python
(t.isnull().mean() * 100).round(1)

U deck fali 77%, u age 20%, u embarked 0.2%. Procenat ti odmah govori šta da radiš: kolonu sa 77% praznih obično izbacuješ, dok se 0.2% popunjava bez razmišljanja.

### Popunjavanje
python
t["age"].fillna(t["age"].median())                # medijana
t["embarked"].fillna(t["embarked"].mode()[0])     # najčešća vrednost

Za age se koristi medijana, a ne prosek, jer je medijana otporna na ekstremne vrednosti. Ovde je medijana 28, a prosek 29.7.

mode() vraća Series, jer može biti više podjednako čestih vrednosti. Zato ide [0] da uzmeš prvu.

Pametnija verzija popunjava po grupama:

python
t["age"] = t["age"].fillna(
    t.groupby(["sex", "pclass"], observed=True)["age"].transform("median")
)

Ovde se spajaju stvari koje već znaš. groupby pravi grupe, a transform("median") vrati medijanu grupe nazad na svaki red te grupe. To je razlika u odnosu na mean(), koji bi vratio 6 brojeva. transform vraća 891, koliko i redova, pa se lepo poklapa sa fillna.

Rezultat ima smisla: muškarac iz prve klase dobija 40 godina, a devojka iz treće 21.5. Bolje nego da svi dobiju istih 28.

Izbacivanje
python
t.dropna()                      # svaki red u kome bilo šta fali
t.dropna(subset=["age"])        # samo gde fali starost
t.dropna(axis=1, thresh=800)    # kolone sa manje od 800 punih vrednosti

Pazi na prvi. Na Titanicu ti od 891 reda ostane samo 182, jer deck obori skoro sve. Zato se dropna() bez argumenata retko koristi. Sa subset=["age"] ostaje 714, što je već upotrebljivo.

### Zamka sa category kolonama
python
t["deck"].fillna("Unknown")

Baca grešku: Cannot setitem on a Categorical with a new category. Kolona je tipa category i zna samo za slova A do G. "Unknown" nije u spisku dozvoljenih.

Rešenje je da je prvo pretvoriš u običan tekst:

python
t["deck"] = t["deck"].astype(str).replace("nan", "Unknown")

Kad pretvoriš u str, NaN postane tekst "nan", pa ga menjaš kao običnu reč.

## 3. Menjanje vrednosti
map: tekst u broj preko rečnika
python
t["sex_num"] = t["sex"].map({"male": 0, "female": 1})

Ovo ti treba stalno, jer modeli ne razumeju reči. Pazi: vrednost koje nema u rečniku postaje NaN. Zato uvek posle proveri sa value_counts(dropna=False).

replace: zamena pojedinačnih vrednosti
python
t["embarked"].replace({"S": "Southampton", "C": "Cherbourg"})

Razlika u odnosu na map: replace ostavlja vrednosti koje nisi naveo, a map ih pretvara u NaN. Zato map koristiš kad prevodiš celu kolonu, a replace kad popravljaš par vrednosti.

astype: promena tipa
python
t["survived"].astype(bool)
t["fare"].astype(int)       # odseca decimale, ne zaokružuje
t["sex"].astype(str)
pd.cut: broj u kategorije
python
t["grupa"] = pd.cut(
    t["age"],
    bins=[0, 12, 18, 60, 100],
    labels=["dete", "tinejdzer", "odrastao", "stariji"],
)

Granice sam biraš. Imaj u vidu da uvek dobiješ jednu granicu više nego oznaka, jer četiri granice prave tri intervala, a pet granica četiri.

Odmah proveri da li nešto znači:

python
t.groupby("grupa", observed=True)["survived"].mean()

Deca 58%, tinejdžeri 43%, odrasli 39%, stariji 23%. Jasan obrazac.

Postoji i pd.qcut, koji sam bira granice tako da u svakoj grupi bude podjednako redova. Koristan za cene karata, gde ti granice nisu očigledne.

.loc za dodelu

Kad menjaš vrednosti samo u nekim redovima, ide .loc sa maskom i imenom kolone:

python
t.loc[t["fare"] > 200, "skupa_karta"] = 1

Čita se: u redovima gde je cena preko 200, u koloni skupa_karta upiši 1.

Nikad ne piši ovako:

python
t[t["fare"] > 200]["skupa_karta"] = 1   # ne radi

Ovo prvo napravi kopiju, upiše u kopiju, pa je baci. Original ostane nepromenjen, a ti dobiješ SettingWithCopyWarning ili tiho ništa. To je ona ista priča o copy i view iz zadatka N5.

python
t["punoletan"] = np.where(t["age"] < 18, "maloletan", "punoletan")

### Za ovo iznad sve znaci imamo par stvari
- value_counts() se koristi za brojanje postojecih vrednosti znaci on nan ne racuna, ako ocemo da prikazemo i koliko ih je nan kazemo u zagradi **dropna = False**. U zagradi stavljamo prvo za koju kolonu prebrojavamo vrednosti, ako hocemo da imam procente onda kazemo u zagradi **normalize = True**
- sort_index() se koristi da se sortira po vrednostima a ne po broju ponavljanja tj frekv tih vrednosti, ako hocemo opadajuce onda stavimo **ascending=False**
- reset_index() se koristi posto nam value_counts vrati Series da to pretvorimo u dataframe odnosno indeksi su nam ustvari vrednosti neke kolone u Series a onda to postaje kolona pored druge kolone koja je frekvenca i to je Dataframe sada sa 2 kolone a indeksi su 0 1 2 itd

### Zadatak: Popuni total_bill medijanom po vremenu obroka, preko groupby i transform.
Znaci krecemo sa dodelom to je logicno posto popunjavamo kolonu, fillna sluzi da popunimo samo vrednosti koje fale, e sad dalje groupby nam je sluzio da grupisemo sve po vremenu za total bill a transform je da se vrate svi redovi odnosno da ne bude jedna vrednost tipa za veceru nego za svaku kolonu gde je vecera ponavljamo tu vrednost jer tako imamo broj redova koliko i u standardnom df pa ce zato da se koristi transform da bi se samo gdee nema vrednost upisala vrednost za taj red iz rezultata

```python
df["total_bill"] = df["total_bill"].fillna(df.groupby("time")["total_bill"].transform("median"))
```
