**Ispod je N3**

```python
ocene = np.array([
    [5, 4, 3, 5],
    [2, 3, 1, 3],
    [5, 5, 4, 4],
    [3, 2, 2, 1],
    [4, 5, 5, 5],
    [1, 3, 2, 4],
])
#axis 0 je po kolonama i to je predmet a axis 1 je po redovima i to je po uceniku
predmeti = np.array(["matematika", "fizika", "hemija", "biologija"])
```

```python
minindeks = np.argmin(ocene.mean(axis = 0))
print(predmeti[minindeks])
```

```python
a = np.any(ocene == 1, axis = 1)
np.count_nonzero(a)
```

```python
np.sum(ocene == 5)
```

```python
niz = ocene.mean(axis = 1)
print(ocene[niz>=4])
```

```python
X = np.array([
    [50, 2, 10],
    [80, 3, 5],
    [120, 4, 20],
    [65, 2, 30],
])  # kvadratura, broj soba, starost stana
w = np.array([1500, 5000, -300])
b = 10000

rez = X@w+b
print(rez)
```

```python
tacke = np.array([[0, 0], [3, 4], [6, 8], [1, 1]])
A = tacke[:, None, :]
B = tacke[None, :, :]

razlike = A-B
d = np.sqrt((razlike ** 2).sum(axis = 2))
print(d)
```
